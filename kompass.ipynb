# Système de Gestion des Employés:

## Introduction:


Le Système de Gestion des Employés Kompass est un système conçu pour gérer les enregistrements des employés en utilisant SQLite et Pandas. Il fournit des fonctionnalités pour ajouter, afficher, mettre à jour et supprimer les informations des employés, ainsi qu'une analyse de données de base en utilisant Pandas.

## 1. Configuration de la Base de Données SQLite:

SQLite est un moteur de base de données léger qui stocke les données dans un fichier unique.

Voici le code pour créer la base de données SQLite et la table d'employés :
Exécutez cette cellule pour créer la base de données et la table.

```python
import sqlite3

# Connexion à la base de données SQLite
conn = sqlite3.connect('kompass_employees.db')
cur = conn.cursor()

# Création de la table employees
cur.execute('''CREATE TABLE IF NOT EXISTS employees (
                    id INTEGER PRIMARY KEY,
                    name TEXT,
                    department TEXT)''')
conn.commit()
```

## 2. Classe de Gestion des Employés:

La classe `EmployeeManagement` gère les opérations CRUD sur les données des employés. Elle interagit avec la base de données SQLite pour effectuer ces opérations.


```python
class EmployeeManagement:
    def __init__(self, db_file):
        self.conn = sqlite3.connect(db_file)
        self.cur = self.conn.cursor()

    def add_employee(self, employee):
        self.cur.execute('''INSERT INTO employees (name, department)
                            VALUES (?, ?)''', (employee['name'], employee['department']))
        self.conn.commit()

    def view_employees(self):
        self.cur.execute("SELECT * FROM employees")
        rows = self.cur.fetchall()
        for row in rows:
            print(row)

    def update_employee(self, employee_id, new_name, new_department):
        self.cur.execute('''UPDATE employees SET name = ?, department = ?
                            WHERE id = ?''', (new_name, new_department, employee_id))
        self.conn.commit()

    def delete_employee(self, employee_id):
        self.cur.execute("DELETE FROM employees WHERE id = ?", (employee_id,))
        self.conn.commit()

    def __del__(self):
        self.conn.close()
```

## 3. Menu Principal:

Le menu principal propose des options pour effectuer diverses opérations sur les données des employés.


```python
def main_menu():
    print("\nSystème de Gestion des Employés Kompass")
    print("1. Ajouter un Employé")
    print("2. Afficher les Employés")
    print("3. Mettre à Jour un Employé")
    print("4. Supprimer un Employé")
    print("5. Quitter")

def gestion():
    emp_management = EmployeeManagement('kompass_employees.db')

    while True:
        main_menu()
        choice = input("Entrez votre choix: ")

        if choice == '1':
            name = input("Entrez le nom de l'employé: ")
            department = input("Entrez le département de l'employé: ")
            emp_management.add_employee({'name': name, 'department': department})
            print("Employé ajouté avec succès.")
        elif choice == '2':
            print("Employés:")
            emp_management.view_employees()
        elif choice == '3':
            employee_id = input("Entrez l'ID de l'employé à mettre à jour: ")
            new_name = input("Entrez le nouveau nom de l'employé: ")
            new_department = input("Entrez le nouveau département de l'employé: ")
            emp_management.update_employee(employee_id, new_name, new_department)
            print("Employé mis à jour avec succès.")
        elif choice == '4':
            employee_id = input("Entrez l'ID de l'employé à supprimer: ")
            emp_management.delete_employee(employee_id)
            print("Employé supprimé avec succès.")
        elif choice == '5':
            print("Quitter...")
            break
        else:
            print("Choix invalide. Veuillez réessayer.")

if __name__ == "__main__":
    gestion()
```
